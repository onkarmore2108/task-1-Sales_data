```python
import pandas as pd 
```

```python
#Load the sales dataset
df = pd.read_csv("sales_data.csv")
```

```python
# show first 5 rows
print("Original Data:")
print(df.head())
```

```python
# 2. Check missing values 
print("\nMissing Values :")
print(df.isnull().sum())
```

```python
# fill missing numerical values with median 

if "Sales_Amount" in df.columns:
    df["Sales_Amount"] = df["Sales_Amount"].fillna(df["Sales_Amount"].median())
```

```python
#fill missing quantity with 1
if "Quantity_Sold"  in df.columns: 
    df["Quantity_Sold "] = df["Quantity_Sold"].fillna(1)
```

```python
#Remove duplicate rows 
df = df.drop_duplicates()
```

```python
#Clean column names 
df.columns = (df.columns.str.strip().str.lower().str.replace("","_"))
```

```python
#Remove extra spaces from text columns 
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()
```

```python
#standardize category
if "Product_Category" in df.columns :
    df["Product_Category"] = df["Product_Category"].str.upper()
```

```python
#convert date columns 
if "Sale_Date" in df.columns:
    df["Sale_Date"] = pd.to_datetime(df["Sale_Date"],errors="coerce",dayfirst=true)
```

```python
# check final data types 
print("\nFinal Data Types: ")
print(df.dtypes)
```

```python
#check misssing values again
print("\nFinal Missing Values: ")
print(df.isnull().sum())
```

```python
# Save cleaned Dataset
df.to_csv("cleaned_sales_data.csv", index=False)
```
